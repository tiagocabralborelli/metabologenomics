```python
from Bio import SeqIO
```

```python
bgc_ = "../BGC/illumina/Scaffold_1.region020.gbk"

with open("Scaffold_1.region020.fasta","w+") as fasta:
    for i in SeqIO.parse(bgc_,"genbank"):
        for featutes in i.features:
            if featutes.type == "CDS":
                fasta.write(f">{featutes.qualifiers.get('ID',[''])[0]}\n{featutes.qualifiers.get('translation',[''])[0]}\n")
```
